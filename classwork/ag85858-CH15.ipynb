```python
import numpy as np
import matplotlib.pyplot as plt

# Function for dy/dx = -2y
def f(x, y):
    return -2 * y

x_start, x_end = 0, 2

# Step sizes from 0.01 to 1 in steps of 0.05
h_values = np.arange(0.01, 1.01, 0.05)

runs = []

plt.figure()

for h in h_values:
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)

    # Initial condition
    y_values[0] = 1

    # Euler's method
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

    runs.append((h, x_values, y_values))
    plt.plot(x_values, y_values, label=f"h = {h:.2f}")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Euler's Method for Different Step Sizes")
plt.legend()
plt.show()

# First run is the reference
first_h, first_x, first_y = runs[0]

for h, x_values, y_values in runs:
    # Compare current run to first run at matching x locations by interpolation
    first_y_interp = np.interp(x_values, first_x, first_y)

    point_errors = np.abs(y_values - first_y_interp)
    total_error = np.sum(point_errors)

    print(f"\nStep size h = {h:.2f}")
    print("Point errors (current run vs first run):")
    for x, err in zip(x_values, point_errors):
        print(f"x = {x:.4f}, error = {err:.6f}")

    print(f"Total accumulated error for h = {h:.2f}: {total_error:.6f}")
    
# *** AI used to clean up code - plots, functions, math, etc. ***
```
