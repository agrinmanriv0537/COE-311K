```python
import numpy as np
import matplotlib.pyplot as plt
import time

def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]

    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]

    return C

sizes = []
naive_times = []
numpy_times = []

size = 3

for matrix_num in range(1, 9):
    A = np.random.rand(size, size)
    B = np.random.rand(size, size)

    start = time.perf_counter()
    naive_matrix_multiplication(A.tolist(), B.tolist())
    end = time.perf_counter()
    naive_times.append(end - start)

    start = time.perf_counter()
    np.matmul(A, B)
    end = time.perf_counter()
    numpy_times.append(end - start)

    sizes.append(matrix_num)
    print(f"Matrix {matrix_num}: {size}x{size}")
    print(f"Naive runtime: {naive_times[-1]:.6f} seconds")
    print(f"Numpy runtime: {numpy_times[-1]:.6f} seconds")
    print()

    size *= 2

plt.plot(sizes, naive_times, marker='o', label='Naive')
plt.plot(sizes, numpy_times, marker='o', label='Numpy')
plt.title('Matrix Number vs Runtime')
plt.xlabel('Matrix Number')
plt.ylabel('Runtime (seconds)')
plt.legend()
plt.show()

# *** AI used to clean up code - plots, functions, math, etc. ***
```
