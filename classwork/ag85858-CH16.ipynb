```python
import numpy as np
import matplotlib.pyplot as plt

k = 1.0
x_start = 0
x_end = 10
y_0 = 1.0
v_0 = 0.0

h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]

plt.figure(figsize=(10, 6))

for h in h_values:
    # Create arrays to store x, y, and v values
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))

    # Set initial conditions
    y_values[0] = y_0
    v_values[0] = v_0

    # Euler's Method
    for i in range(1, len(x_values)):
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]

        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)

    plt.plot(x_values, y_values, label=f'h = {h}')

plt.xlabel('x')
plt.ylabel('y')
plt.title("Euler's Method for d²y/dx² = -ky with different h values")
plt.legend()
plt.grid(True)
plt.show()


# Part 2
k = 10
h = 0.01

x_values = np.arange(x_start, x_end + h, h)
y_values = np.zeros(len(x_values))
v_values = np.zeros(len(x_values))

y_values[0] = y_0
v_values[0] = v_0

for i in range(1, len(x_values)):
    x_n = x_values[i - 1]
    y_n = y_values[i - 1]
    v_n = v_values[i - 1]

    y_values[i] = y_n + h * v_n
    v_values[i] = v_n + h * (-k * y_n)

    
# AI Used for nice graphing
    
plt.figure(figsize=(10, 5))
plt.plot(x_values, y_values, label='Position y(x)')
plt.plot(x_values, v_values, label='Velocity v(x)', linestyle='--')
plt.xlabel('x')
plt.ylabel('y and v')
plt.title("Solution of d²y/dx² = -10y using Euler's Method")
plt.legend()
plt.grid(True)
plt.show()
```

## Writeup: 
 - Becomes unstable around h≈0.2 and larger.
 - Yes, instability shows as growing oscillation amplitude.
 - Larger k makes oscillations faster and instability appear sooner.
 - For k=10,h=0.01, solution is stable but slightly drifts.
 - Euler’s Method is not reliable for large h on oscillatory systems.
