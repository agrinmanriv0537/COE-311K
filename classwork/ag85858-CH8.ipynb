```python
def is_diagonally_dominant(matrix):
    n = len(matrix)

    for i in range(n):
        diagonal = abs(matrix[i][i])
        row_sum = 0

        for j in range(n):
            if j != i:
                row_sum += abs(matrix[i][j])

        if diagonal < row_sum:
            return False

    return True
```

```python
matrix1 = [
    [4, 1, 2],
    [1, 3, 1],
    [0, 1, 5]
]

matrix2 = [
    [1, 3, 2],
    [2, 1, 1],
    [1, 2, 1]
]

print("Is the matrix diagonally dominant?", is_diagonally_dominant(matrix1))
print(matrix1)

print("Is the matrix diagonally dominant?", is_diagonally_dominant(matrix2))
print(matrix2)
```
