```python
import numpy as np
import time

EPS = 1e-6

def jacobi(A, b, eps=EPS, max_iter=100000):
    n = len(b)
    x = np.zeros(n)

    start = time.perf_counter()
    for k in range(1, max_iter + 1):
        x_new = np.zeros(n)
        for i in range(n):
            s = 0.0
            for j in range(n):
                if j != i:
                    s += A[i, j] * x[j]
            x_new[i] = (b[i] - s) / A[i, i]

        if np.linalg.norm(x_new - x, 2) < eps:
            return x_new, k, time.perf_counter() - start, True

        x = x_new

    return x, max_iter, time.perf_counter() - start, False


def gauss_seidel(A, b, eps=EPS, max_iter=100000):
    n = len(b)
    x = np.zeros(n)

    start = time.perf_counter()
    for k in range(1, max_iter + 1):
        x_old = x.copy()

        for i in range(n):
            s1 = 0.0
            for j in range(i):         # uses newest values
                s1 += A[i, j] * x[j]

            s2 = 0.0
            for j in range(i + 1, n):  # uses previous-iteration values
                s2 += A[i, j] * x_old[j]

            x[i] = (b[i] - s1 - s2) / A[i, i]

        if np.linalg.norm(x - x_old, 2) < eps:
            return x, k, time.perf_counter() - start, True

    return x, max_iter, time.perf_counter() - start, False


matrixAone = np.array([

[0.582745, 0.48 , 0.10 , 0.  , 0.  ],
[0.48 , 1.044129, 0.46 , 0.10 , 0.  ],
[0.10 , 0.46 , 1.10431 , 0.44 , 0.10 ],
[0.  , 0.10 , 0.44 , 0.963889, 0.42 ],
[0.  , 0.  , 0.10 , 0.42 , 0.522565]], dtype=float)

matrixBone = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)

matrixAtwo = np.array([[1, 1], [1, 1.0001]])
matrixBtwo = np.array([1, 1])

matrixAthree = np.array([[1, 1], [1, 1.0001]])
matrixBthree = np.array([1, 1.0001])


def run_case(label, A, b):
    xj, kj, tj, okj = jacobi(A, b, eps=EPS)
    xg, kg, tg, okg = gauss_seidel(A, b, eps=EPS)

    print(f"\nChallenge {label}")
    print("  Jacobi:", "converged" if okj else "DID NOT converge",
          "| iters =", kj, "| time =", tj, "| x =", xj)
    print("  G-Seidel:", "converged" if okg else "DID NOT converge",
          "| iters =", kg, "| time =", tg, "| x =", xg)


run_case("One:", matrixAone, matrixBone)
run_case("Two:", matrixAtwo, matrixBtwo)
run_case("Three:", matrixAthree, matrixBthree)
```
